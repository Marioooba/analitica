```python
import pandas as pd
import numpy as np


# Cargar el archivo CSV
df = pd.read_csv('vgratings2017.csv')

# Inspección inicial
print("Primeras filas del conjunto de datos:")
df.head()
```

```python
df.info()
df.describe()
```

```python
# Revisión de valores faltantes
print(df.isnull().sum())
```

```python
# Si hay duplicados, eliminarlos
df = df.drop_duplicates()
```

```python

#Numericas
#Year_Of_Release, 5   NA_Sales   EU_Sales JP_Sales Other_Sales  Global_Sales Critic_Score Critic_Count User_Score User_Count       

df['Year'] = df['Year'].fillna(df['Year'].median())
df['NA_Sales'] = df['NA_Sales'].fillna(df['NA_Sales'].median())
df['EU_Sales'] = df['EU_Sales'].fillna(df['EU_Sales'].median())
df['JP_Sales'] = df['JP_Sales'].fillna(df['JP_Sales'].median())
df['Other_Sales'] = df['Other_Sales'].fillna(df['Other_Sales'].median())
df['Global_Sales'] = df['Global_Sales'].fillna(df['Global_Sales'].median())

#Criticas
df['Critic_Score'] = df['Critic_Score'].fillna(df['Critic_Score'].median())
df['Critic_Count'] = df['Critic_Count'].fillna(df['Critic_Count'].median())
df['User_Score'] = df['User_Score'].fillna(df['User_Score'].median())
df['User_Count'] = df['User_Count'].fillna(df['User_Count'].median())
   

#Categoricas
#Publisher, Rating

df['Publisher'] = df['Publisher'].fillna(df['Publisher'].mode()[0])
df['Rating'] = df['Rating'].fillna(df['Rating'].mode()[0])
```

```python
#Checamos los datos limpiados
print(df.isnull().sum())
```

```python
# Guardar los datos limpios en un nuevo archivo CSV
df.to_csv('clean_games.csv', index=False)
```
